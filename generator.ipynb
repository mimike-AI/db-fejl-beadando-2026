```python
from faker import Faker

from data.model_dataclasses import Person, Sex
import random


def generate_people(n: int, male_ratio: float = 0.5,
                    locale: str = "hu_HU",
                    unique: bool = False,
                    min_age: int = 0,
                    max_age: int = 100) -> list[Person]:

    assert n > 0
    assert 0 <= male_ratio <= 1
    assert min_age >= 0
    assert min_age <= max_age <= 100

    people = []
    fake = Faker(locale)
    # Faker.seed(0)
    fake = fake if not unique else fake.unique
    for i in range(n):
        sex = Sex.MALE if random.random() < male_ratio else Sex.FEMALE
        person = Person(
            f"P-{str(i + 1).zfill(6)}",
            fake.unique.name_male() if sex is Sex.MALE else fake.unique.name_female(),
            random.randint(min_age, max_age),
            sex)

        people.append(person)

    return people
```

```python
for person in generate_people(12):
    print(person)
```
