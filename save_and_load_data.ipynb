Minden típus összekapcsolása:

````python
person_by_id = {p.id: p for p in people}
health_by_person = {h.person_id: h for h in profiles}
diagnoses_by_person = {}
for d in diagnoses:
    diagnoses_by_person.setdefault(d.person_id, []).append(d)

for galactic in galactic_profiles:
    person = person_by_id[galactic.person_id]
    health = health_by_person[galactic.person_id]
    person_diagnoses = diagnoses_by_person.get(galactic.person_id, [])

    print(f"{galactic.rank} {galactic.scifi_name} ({person.name}, {person.age}) – {galactic.starship}")
    print(f"    vércsoport: {health.blood_type}, {health.height_cm} cm, {health.weight_kg} kg, BMI {health.bmi}")
    for d in person_diagnoses:
        print(f"    diagnózis: {d.disease} ({d.icd10}, {d.specialty})")
    if not person_diagnoses:
        print("    diagnózis: nincs")
````

A kapcsolatok adatbázis-szemmel:

  Person (id) ──1:1── GalacticProfile (person_id)
              ──1:1── HealthProfile   (person_id)
              ──1:N── Diagnosis       (person_id)

  A person_id itt idegen kulcs, a cella pedig egy JOIN-t végez el Pythonban.

  1. Három szótár (dict) készül, amiknél a person_id a kulcs. Ezekből egy lépésben elő lehet venni egy személy adatait, ami gyors keresést eredményez.
     - person_by_id és health_by_person: egy kulcshoz egy érték tartozik, ez az 1:1 kapcsolat.
     - diagnoses_by_person: egy kulcshoz egy lista tartozik, ez az 1:N kapcsolat. A setdefault(kulcs, []) üres listát hoz létre, amíg a kulcs még nincs a szótárban, majd visszaadja a listát, amihez aztán az append hozzáfűzi a diagnózist.
  2. A galaktikus profilokon megy végig a ciklus, és mindegyikhez kikeresi a hozzá tartozó személyt, annak az egészségügyi profilját és az esetleges diagnózisokat.
  3. A diagnózisoknál .get(..., [])-t használtam, nem [...]-t. Mivel van, akinek nincs diagnózisa, és a [...] ilyenkor KeyError-t dobna. Ez a LEFT JOIN megfelelője: a szereplő akkor is megjelenik, ha nincs hozzá diagnózis.

  SQL-ben ugyanez így nézne ki:
```sql
SELECT g.rank, g.scifi_name, p.name, h.blood_type, d.disease
FROM galactic_profiles g
JOIN people p           ON p.id = g.person_id
JOIN health_profiles h  ON h.person_id = g.person_id
LEFT JOIN diagnoses d   ON d.person_id = g.person_id;
```
