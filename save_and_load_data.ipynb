```python
from openpyxl import load_workbook
from data.handler.xlsx_handler import read_people, read_health_profiles, read_diagnoses, read_galactic_profiles
wb = load_workbook("patient.xlsx")
print(wb.sheetnames)

loaded = read_people(wb)
for person in loaded:
    print(person)
print(loaded == people)

loaded_profiles = read_health_profiles(wb)
for profile in loaded_profiles:
    print(profile)
print(loaded_profiles == profiles)

loaded_diagnoses = read_diagnoses(wb)
for diagnosis in loaded_diagnoses:
    print(diagnosis)
print(loaded_diagnoses == diagnoses)

loaded_galactic = read_galactic_profiles(wb)
for galactic in loaded_galactic:
    print(galactic)
print(loaded_galactic == galactic_profiles)
```
